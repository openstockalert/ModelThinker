# Chapter 15 (part b) — Local interaction models

Companion notebook to `docs/models/ch15b_life_and_majority.md` and the Streamlit page `app/pages/15b_Life_and_Majority.py`.

Two cellular automata sharing the same 'update from Moore neighbours' mechanic — but producing wildly different global patterns:
- **Conway's Game of Life** — 3 rules → oscillators, gliders, chaos.
- **Local Majority Model** — 1 rule → the grid freezes into stable clusters.

```python
import pathlib
import sys

sys.path.insert(0, str(pathlib.Path.cwd().parent / 'src'))

import matplotlib.pyplot as plt

from modelthinker.dynamics.life_and_majority import (
    count_clusters,
    place_pattern,
    simulate_life,
    simulate_majority,
)
```

## 1. Game of Life — the glider crossing a torus

```python
grid = place_pattern('glider', grid_size=24, offset_r=2, offset_c=2)
r = simulate_life(initial_grid=grid, steps=80, wrap=True, stop_when_stable=False)

fig, axes = plt.subplots(1, 5, figsize=(15, 3))
for ax, t in zip(axes, [0, 4, 12, 24, 60], strict=True):
    ax.imshow(r.grid_history[t], cmap='Greys', vmin=0, vmax=1)
    ax.set_title(f't = {t}')
    ax.set_xticks([]); ax.set_yticks([])
plt.tight_layout(); plt.show()
print('Live cells at each of these steps: 5 for all snapshots (glider preserves cell count).')
```

## 2. Game of Life — the R-pentomino methuselah

Five cells. Over 1000 steps of chaos. This is the pattern that convinced Conway the game was interesting.

```python
grid = place_pattern('r_pentomino', grid_size=80, offset_r=40, offset_c=40)
r = simulate_life(initial_grid=grid, steps=1200, wrap=False, stop_when_stable=True)

fig, ax = plt.subplots(figsize=(9, 3))
ax.plot(r.populations, color='#4C6EF5', lw=1.5)
ax.set(xlabel='step', ylabel='live cells',
       title=f'R-pentomino population over {r.steps_taken} steps '
             f'(stable={r.stable})')
ax.grid(alpha=0.3); plt.show()
print(f'Peak population: {max(r.populations)} at step {r.populations.index(max(r.populations))}')
```

## 3. Local Majority — random start → clustered end

```python
r = simulate_majority(grid_size=50, initial_density=0.5, steps=200, seed=7)

fig, axes = plt.subplots(1, 4, figsize=(14, 4))
for ax, t in zip(axes, [0, 2, 6, r.steps_taken], strict=True):
    ax.imshow(r.grid_history[t], cmap='coolwarm', vmin=0, vmax=1)
    ax.set_title(f't = {t}')
    ax.set_xticks([]); ax.set_yticks([])
plt.tight_layout(); plt.show()
print(f'Stable at step {r.steps_taken}: {r.stable}')
print(f'Final 1-clusters: {count_clusters(r.grid_history[-1], value=1)}')
```
