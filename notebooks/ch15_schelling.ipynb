# Chapter 15 — Schelling segregation

Companion notebook to `docs/models/ch15_schelling.md`. Individually mild same-type preferences on a grid produce dramatic collective segregation.


```python
import pathlib
import sys

sys.path.insert(0, str(pathlib.Path.cwd().parent / 'src'))

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import ListedColormap

from modelthinker.dynamics.schelling import simulate

CMAP = ListedColormap(['white', '#4C6EF5', '#F76707'])
```

## 1. The classic result: τ = 0.3

Everyone is happy being a 30 % minority. Yet clusters emerge anyway.

```python
r = simulate(grid_size=40, density=0.9, tolerance=0.3, max_steps=8000, snapshot_every=200, seed=42)

fig, axes = plt.subplots(1, 4, figsize=(14, 4))
picks = [0, len(r.grid_history)//3, 2*len(r.grid_history)//3, len(r.grid_history)-1]
for ax, i in zip(axes, picks, strict=True):
    ax.imshow(r.grid_history[i], cmap=CMAP, vmin=0, vmax=2)
    ax.set_title(f'snapshot {i}/{len(r.grid_history)-1}\nseg = {r.seg_index_history[i]:.2f}')
    ax.set_xticks([]); ax.set_yticks([])
plt.tight_layout(); plt.show()
print(f'Converged: {r.converged}, steps: {r.steps_taken}')
```

## 2. Tolerance sweep — the phase transition

Sweep τ and record final segregation.

```python
taus = np.linspace(0.0, 0.8, 9)
finals = []
for tau in taus:
    r = simulate(grid_size=25, density=0.9, tolerance=float(tau), max_steps=6000, snapshot_every=500, seed=7)
    finals.append(r.seg_index_history[-1])

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(taus, finals, 'o-', color='#4C6EF5', lw=2)
ax.axhline(0.5, color='gray', ls=':', label='random baseline')
ax.set(xlabel=r'tolerance $\tau$', ylabel='final segregation index',
       title='Segregation vs individual tolerance', ylim=(0, 1))
ax.legend(); plt.show()
```

## 3. What-ifs to try yourself

1. Density trap: set `density=0.98`, `tolerance=0.4`. Watch unhappy agents unable to relocate.
2. Try `tolerance=0.7` — the model can't converge, but segregation doesn't rise much either.
3. Very small grid (`grid_size=10`): boundary effects dominate and results become noisy.
