# Chapter 11 — SIR epidemic (contagion)

Companion notebook to `docs/models/ch11_sir.md`. Three compartments, two rates, one number R₀.

```python
import pathlib
import sys

sys.path.insert(0, str(pathlib.Path.cwd().parent / 'src'))

import matplotlib.pyplot as plt
import numpy as np

from modelthinker.networks.sir import simulate
```

## 1. Baseline outbreak — R₀ = 3

```python
r = simulate(beta=0.3, gamma=0.1, N=100_000, I0=10, days=180)

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(r.t, r.S, label='S', color='#4C6EF5', lw=2)
ax.plot(r.t, r.I, label='I', color='#E03131', lw=2)
ax.plot(r.t, r.R, label='R', color='#37B24D', lw=2)
ax.axvline(r.peak_day, color='gray', ls=':')
ax.set(xlabel='day', ylabel='people', title=f'SIR — R₀ = {r.R0:.1f}, peak day {r.peak_day:.0f}')
ax.legend(); plt.show()
print(f'peak I: {r.peak_I:,.0f} | final R: {r.final_R:,.0f} ({100*r.final_R/r.N:.1f}% infected ever)')
```

## 2. R₀ threshold — sweep β with γ fixed

```python
betas = np.linspace(0.05, 0.5, 10)
attack_rates = []
for b in betas:
    r = simulate(beta=float(b), gamma=0.1, N=100_000, I0=10, days=500)
    attack_rates.append(r.final_R / r.N)

R0s = betas / 0.1
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(R0s, attack_rates, 'o-', color='#4C6EF5', lw=2)
ax.axvline(1, color='red', ls='--', label='R₀ = 1 threshold')
ax.set(xlabel='R₀', ylabel='final attack rate (fraction ever infected)',
       title='The R₀ threshold is a phase transition')
ax.legend(); ax.grid(alpha=0.3); plt.show()
```
