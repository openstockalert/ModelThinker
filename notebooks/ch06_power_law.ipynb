# Chapter 6 — Power laws (Barabási–Albert)

Companion notebook to `docs/models/ch06_power_law.md`. Preferential attachment produces power-law degree distributions.

```python
import pathlib
import sys

sys.path.insert(0, str(pathlib.Path.cwd().parent / 'src'))

import matplotlib.pyplot as plt

from modelthinker.distributions.power_law import degree_pmf, simulate
```

## 1. Log-log degree distribution

```python
r = simulate(n=5000, m=2, seed=42)
ks, pk = degree_pmf(r.degrees)

fig, ax = plt.subplots(figsize=(7, 4))
ax.loglog(ks, pk, 'o', color='#4C6EF5', label='empirical')
ks_tail = ks[ks >= r.m]
fit = pk[ks >= r.m][0] * (ks_tail / ks_tail[0]) ** (-r.estimated_exponent)
ax.loglog(ks_tail, fit, 'r--', label=f'fit: k^-{r.estimated_exponent:.2f}')
ax.set(xlabel='degree k', ylabel='P(k)', title=f'BA n={r.n}, m={r.m}')
ax.legend(); ax.grid(True, which='both', alpha=0.3); plt.show()
```

## 2. Lorenz curve — how concentrated is degree?

```python
fig, ax = plt.subplots(figsize=(6, 6))
ax.fill_between(r.lorenz_x, r.lorenz_y, alpha=0.3, color='#4C6EF5')
ax.plot(r.lorenz_x, r.lorenz_y, color='#4C6EF5', lw=2, label='Lorenz')
ax.plot([0, 1], [0, 1], 'k:', label='perfect equality')
ax.set(xlabel='fraction of nodes', ylabel='fraction of degree',
       title=f'Gini = {r.gini:.3f}')
ax.legend(); ax.set_aspect('equal'); plt.show()
```
