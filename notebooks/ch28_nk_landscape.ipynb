# Chapter 28 — NK rugged landscapes

Companion notebook to `docs/models/ch28_nk_landscape.md`. Interacting decisions create rugged landscapes where local hill climbing gets stuck.

```python
import pathlib
import sys

sys.path.insert(0, str(pathlib.Path.cwd().parent / 'src'))

import matplotlib.pyplot as plt
import numpy as np

from modelthinker.learning.nk_landscape import NKLandscape, compare_strategies, count_local_peaks
```

## 1. Strategy comparison on smooth vs rugged

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4), sharey=True)
for ax, K in zip(axes, [0, 8], strict=True):
    res = compare_strategies(N=12, K=K, budget=500, n_restarts=10, long_jump_p=0.2, seed=42)
    for name, r in res.items():
        ax.plot(r.trajectory, label=name, lw=2)
    ax.set(title=f'N=12, K={K}', xlabel='evaluations', ylabel='best fitness so far')
    ax.legend(); ax.grid(alpha=0.3)
plt.tight_layout(); plt.show()
```

## 2. Number of local peaks vs K

```python
N = 10
Ks = list(range(0, N))
avg_peaks = []
for K in Ks:
    counts = [count_local_peaks(NKLandscape(N=N, K=K, seed=s)) for s in range(5)]
    avg_peaks.append(np.mean(counts))

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(Ks, avg_peaks, 'o-', color='#4C6EF5', lw=2)
ax.set(xlabel='K', ylabel='mean local peaks (5 seeds)',
       title=f'Ruggedness explodes with K (N={N})')
ax.grid(alpha=0.3); plt.show()
```
